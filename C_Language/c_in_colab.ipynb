# Running C in Google Colab

1. Run the first code cell to write the C file.
2. Run the second cell to compile and run it.

Keep `%%writefile` as the very first line of the cell.

```python
%%writefile hello.c
#include <stdio.h>

int main() {
    int n = 5;
    printf("Hello from C in Colab!\n");
    for (int i = 1; i <= n; i++) {
        printf("%d squared = %d\n", i, i * i);
    }
    return 0;
}
```

```python
!gcc hello.c -o hello && ./hello
```
